```python
import random

# Fitness function
def fitness(x):
    return x * x


# Create initial population
population_size = 10
population = [random.randint(0, 31) for _ in range(population_size)]

generations = 20

for generation in range(generations):

    # Calculate fitness
    population.sort(key=fitness, reverse=True)

    # Select the best two individuals
    parent1 = population[0]
    parent2 = population[1]

    # Crossover
    child1 = (parent1 + parent2) // 2
    child2 = random.randint(min(parent1, parent2),
                            max(parent1, parent2))

    # Mutation
    if random.random() < 0.1:
        child1 = random.randint(0, 31)

    if random.random() < 0.1:
        child2 = random.randint(0, 31)

    # Create new population
    population = [
        parent1,
        parent2,
        child1,
        child2
    ]

    # Fill remaining population
    while len(population) < population_size:
        population.append(random.randint(0, 31))

    print(
        "Generation:", generation + 1,
        "Best:", population[0],
        "Fitness:", fitness(population[0])
    )


# Final result
best = max(population, key=fitness)

print("\nOptimal Solution:", best)
print("Maximum Fitness:", fitness(best))
```
