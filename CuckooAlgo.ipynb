```python
import random

# Fitness function
def fitness(x):
    return x * x

# Initialize parameters
n = 3
Pa = 0.25
Max = 20

# Generate random nests
nests = [random.uniform(-10, 10) for i in range(n)]

# Cuckoo Search
for i in range(Max):

    # Select a random nest
    j = random.randint(0, n-1)

    # Generate new solution using Levy-like random step
    new = nests[j] + random.uniform(-2, 2)

    # Evaluate and replace if better
    if fitness(new) < fitness(nests[j]):
        nests[j] = new

    # Abandon worst nest
    if random.random() < Pa:
        k = max(range(n), key=lambda x: fitness(nests[x]))
        nests[k] = random.uniform(-10, 10)

# Find best solution
best = min(nests, key=fitness)
```

```python
print(f"Best solution found: {best}")
print(f"Fitness value: {fitness(best)}")
```
