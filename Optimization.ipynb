```python
import random
import math

# Target function: y = x^2 + 2x + 1
def target_function(x):
    return x**2 + 2*x + 1


# Available mathematical functions
def add(a, b):
    return a + b

def sub(a, b):
    return a - b

def mul(a, b):
    return a * b


functions = [add, sub, mul]
function_symbols = ['+', '-', '*']
terminals = ['x', '1', '2']


# Generate a random expression
def generate_expression(depth=0, max_depth=3):
    if depth >= max_depth or random.random() < 0.3:
        return random.choice(terminals)

    func = random.randint(0, len(functions) - 1)

    left = generate_expression(depth + 1, max_depth)
    right = generate_expression(depth + 1, max_depth)

    return f"({left} {function_symbols[func]} {right})"


# Evaluate an expression
def evaluate_expression(expression, x):
    try:
        expression = expression.replace('x', str(x))
        return eval(expression)
    except:
        return float('inf')


# Calculate fitness
def fitness(expression):
    error = 0

    for x in range(-5, 6):
        actual = target_function(x)
        predicted = evaluate_expression(expression, x)

        error += abs(actual - predicted)

    return error


# Create initial population
population_size = 50
population = [
    generate_expression()
    for _ in range(population_size)
]


# Evolution process
generations = 100

for generation in range(generations):

    # Sort according to fitness
    population.sort(key=fitness)

    # Best solution
    best = population[0]

    print(
        "Generation:",
        generation + 1,
        " Best:",
        best,
        " Fitness:",
        fitness(best)
    )

    # Stop if a perfect solution is found
    if fitness(best) == 0:
        break

    # Select best individuals
    survivors = population[:10]

    # Create new population
    new_population = survivors.copy()

    while len(new_population) < population_size:

        parent = random.choice(survivors)

        # Mutation
        child = generate_expression()

        # Occasionally keep part of the parent
        if random.random() < 0.5:
            child = parent

        new_population.append(child)

    population = new_population


print("\nFinal Optimized Expression:")
print(best)

print("Fitness:", fitness(best))
```
